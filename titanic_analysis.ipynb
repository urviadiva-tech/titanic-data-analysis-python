**DATA EXTRACTING:**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
df = pd.read_csv("titanic.train.csv")

# Viewing first 5 rows
df.head()
```

```python
# Shape of dataset
df.shape

# Info about columns, datatypes, and null values
df.info()
```

**DATA CLEANING:**

 Handling missing values and duplicates

```python
# Check missing values
df.isnull().sum()
```

```python
# Fill all missing values with 0
df = df.fillna(0)
```

```python
# Drop duplicates if any
df = df.drop_duplicates()
```

```python
# Removing rows where every column is NaN, but keeps rows with at least one valid value.
df.dropna(how='all', inplace=True)
```

```python
# Check again
print("Shape after cleaning:", df.shape)
df.isnull().sum()
```

```python
# Viewing the entire dataframe
df
```

```python
df.info() #information after cleaning
```

**DATA VISUALIZATION:**

1. Visualization with Matplotlib

```python
import matplotlib.pyplot as plt

# 1. Line Plot - example: PassengerId vs Age
plt.figure(figsize=(8,5))
plt.plot(df['PassengerId'], df['Age'], color='seagreen')
plt.xlabel("Passenger ID")
plt.ylabel("Age")
plt.title("Line Plot - Passenger ID vs Age")
plt.show()
```

```python
# 2. Bar plot
plt.figure(figsize=(8,5))
df['Survived'].value_counts().plot(kind='bar', color=['magenta','red'])
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.title("Bar Plot - Survival Counts (Matplotlib)")
plt.show()
```

```python
# 3. Histogram - example: Age distribution
plt.figure(figsize=(8,5))
plt.hist(df['Age'], bins=40, color='blue', edgecolor='black')
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Histogram - Age Distribution (Matplotlib)")
plt.show()
```

```python
# 4. Scatter Plot - example: Age vs Fare
plt.figure(figsize=(10,6))
plt.scatter(df['Age'], df['Fare'], marker='o', color='yellow', alpha=0.6, label='Circle Marker')
plt.scatter(df['Age'], df['Fare'], marker='x', color='red', alpha=0.6, label='Triangle Marker')
plt.title("Age vs Fare with Different Markers")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend()
plt.show()
```

```python
# 5. Pie Chart - example: Male vs Female
plt.figure(figsize=(8,5))
df['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%', startangle=90, colors=['cyan','yellow'])
plt.title("Pie Chart - Gender Distribution")
plt.ylabel("")  # remove y-label
plt.show()
```

2. Visualization with Seaborn

```python
import seaborn as sns

# 1. Line Plot
plt.figure(figsize=(8,5))
sns.lineplot(x='PassengerId', y='Age', data=df, color='blue')
plt.title("Line Plot - Passenger ID vs Age")
plt.show()
```

```python
# 2. Bare Plot
plt.figure(figsize=(6,4))
sns.barplot(x='Survived', hue='Sex', data=df, palette='Set1')  # Use 'Set1' palette
plt.xlabel("Survived")
plt.ylabel("Count")
plt.title("Bar Plot - Survival Counts by Gender")
plt.show()
```

```python
# 3. Histogram (as distribution plot)
plt.figure(figsize=(6,4))
sns.histplot(df['Age'], bins=20, kde=True, color='lightpink')
plt.title("Histogram - Age Distribution")
plt.show()
```

```python
# 4. Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(x='Age', y='Fare', data=df, hue='Sex')
plt.title("Scatter Plot - Age vs Fare")
plt.show()
```

Visualizing outliers and correlation using seaborn:

```python
# Boxplot to detect outliers in Age
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Age'])
plt.title("Boxplot - Outliers in Age")
plt.show()
```

```python
#correlation visualization
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```
